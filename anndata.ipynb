```python
import numpy as np
import anndata as ad
```

```python
adata = ad.read_h5ad("data/hcd_subset.h5ad")
print(adata)
print(adata.uns.get("log1p"))
```

```python
print("X shape:", adata.X.shape)
print("Counts shape:", adata.layers["counts"].shape)

print("X range:", np.min(adata.X), np.max(adata.X))
print("Counts range:",
      np.min(adata.layers["counts"]),
      np.max(adata.layers["counts"]))
```

```python
print("Donors:", adata.obs["donor"].nunique())
print("Cell types:", adata.obs["cell_type"].nunique())
print("Cytokines:", adata.obs["cytokine"].unique())

print("\nCells per condition:")
print(adata.obs.groupby(
    ["donor", "cell_type", "cytokine"],
    observed=True
).size().describe())
```
